```python
import numpy as np
import math

# 1. Load the Data
# Marks for 50 randomly selected students from each section
section_a = [
    8, 9, 10, 11, 12, 10, 11, 12, 13, 14,
    11, 12, 13, 14, 15, 12, 13, 14, 15, 16,
    13, 14, 15, 16, 17, 14, 15, 16, 17, 18,
    15, 16, 17, 18, 19, 16, 17, 18, 19, 20,
    17, 18, 19, 20, 21, 18, 19, 20, 21, 22
]

section_b = [
    15, 16, 17, 18, 19, 17, 18, 19, 20, 21,
    18, 19, 20, 21, 22, 19, 20, 21, 22, 23,
    20, 21, 22, 23, 24, 21, 22, 23, 24, 25,
    22, 23, 24, 25, 26, 23, 24, 25, 26, 27,
    24, 25, 26, 27, 28, 25, 26, 27, 28, 29
]

# Prior probabilities given in the problem statement
P_A = 0.5
P_B = 0.5

# The new student's mark
x = 20

# 2. Calculate Mean and Standard Deviation for both sections
# Using ddof=1 for sample standard deviation to be statistically rigorous
mean_a = np.mean(section_a)
std_a = np.std(section_a, ddof=1)

mean_b = np.mean(section_b)
std_b = np.std(section_b, ddof=1)

# 3. Define the Gaussian (Normal) Density Function
def gaussian_pdf(x, mean, std):
    """Calculates the Gaussian probability density for a given value."""
    exponent = math.exp(-((x - mean) ** 2) / (2 * std ** 2))
    return (1 / (std * math.sqrt(2 * math.pi))) * exponent

# 4. Calculate Likelihoods: P(x=20 | Section)
likelihood_a = gaussian_pdf(x, mean_a, std_a)
likelihood_b = gaussian_pdf(x, mean_b, std_b)

# 5. Apply Bayes' Theorem
# We calculate the proportional posterior: P(x | C) * P(C)
posterior_a = likelihood_a * P_A
posterior_b = likelihood_b * P_B

# 6. Output the Results
print(f"--- Statistical Analysis ---")
print(f"Section A: Mean = {mean_a:.2f}, Standard Deviation = {std_a:.2f}")
print(f"Section B: Mean = {mean_b:.2f}, Standard Deviation = {std_b:.2f}\n")

print(f"--- Likelihoods ---")
print(f"P(X=20 | Section A) = {likelihood_a:.6f}")
print(f"P(X=20 | Section B) = {likelihood_b:.6f}\n")

print(f"--- Bayes' Posterior Probabilities (Proportional) ---")
print(f"Proportional P(Section A | X=20) = {posterior_a:.6f}")
print(f"Proportional P(Section B | X=20) = {posterior_b:.6f}\n")

print(f"--- Final Classification ---")
if posterior_a > posterior_b:
    print("Conclusion: The student is more likely to belong to Section A.")
elif posterior_b > posterior_a:
    print("Conclusion: The student is more likely to belong to Section B.")
else:
    print("Conclusion: The student has an equal probability of belonging to either section.")
```
